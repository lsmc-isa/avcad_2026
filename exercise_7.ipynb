### Análise da Distribuição das Variáveis e Necessidade de Transformação

Para as análises de regressão, é importante entender a distribuição das variáveis, especialmente se são contínuas. Variáveis com distribuições muito assimétricas podem necessitar de transformação (e.g., logarítmica) para melhorar o desempenho do modelo e atender aos pressupostos de normalidade dos resíduos. Vamos visualizar a distribuição das variáveis `Total_sp` (riqueza de espécies) e das variáveis ambientais contínuas mencionadas:

*   `Altitude`
*   `Actual_river_slope`
*   `Elevation_mean_catch`
*   `prec_ann_catch`
*   `temp_ann`
*   `temp_jan`
*   `temp_jul`

```python
continuous_vars = [
    'Total_sp',
    'Altitude',
    'Actual_river_slope',
    'Elevation_mean_catch',
    'prec_ann_catch',
    'temp_ann',
    'temp_jan',
    'temp_jul'
]

# Criar subplots para as distribuições
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(18, 15))
axes = axes.flatten() # Achatar o array de eixos para fácil iteração

for i, var in enumerate(continuous_vars):
    if var in df.columns:
        sns.histplot(df[var], kde=True, ax=axes[i], color='skyblue')
        axes[i].set_title(f'Distribuição de {var}')
        axes[i].set_xlabel(var)
        axes[i].set_ylabel('Frequência')
    else:
        axes[i].set_title(f'Variável {var} não encontrada')
        axes[i].axis('off') # Desliga o eixo se a variável não for encontrada

# Esconder eixos vazios, se houver
for j in range(i + 1, len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.show()
```

### Aplicação de Transformações em Variáveis

Com base na análise das distribuições, vamos aplicar transformações logarítmicas (usando `np.log1p` para lidar com valores zero ou próximos de zero) nas seguintes variáveis para tentar normalizar suas distribuições:

*   `Total_sp`
*   `Actual_river_slope`
*   `prec_ann_catch`

Serão criadas novas colunas no DataFrame com o sufixo `_log` para as variáveis transformadas.

```python
import numpy as np

# Variáveis a serem transformadas (usando log1p para lidar com zeros)
df['Total_sp_log'] = np.log1p(df['Total_sp'])
df['Actual_river_slope_log'] = np.log1p(df['Actual_river_slope'])
df['prec_ann_catch_log'] = np.log1p(df['prec_ann_catch'])

print("Visualização das primeiras linhas do DataFrame com as novas variáveis transformadas:")
display(df[['Total_sp', 'Total_sp_log', 'Actual_river_slope', 'Actual_river_slope_log', 'prec_ann_catch', 'prec_ann_catch_log']].head())
```

### 2. Regressão Linear Múltipla

Agora, vamos construir um modelo de regressão linear múltipla utilizando todas as variáveis preditoras contínuas (e suas versões transformadas, onde aplicável) que foram analisadas individualmente. O objetivo é entender como elas, em conjunto, explicam a riqueza de espécies e observar as mudanças nos coeficientes em comparação com os modelos univariados. Também serão gerados os gráficos de dependência parcial para visualizar os efeitos individuais de cada preditor no contexto do modelo múltiplo.

As variáveis preditoras incluídas serão:

*   `Altitude`
*   `Actual_river_slope_log`
*   `Elevation_mean_catch`
*   `prec_ann_catch_log`
*   `temp_ann`
*   `temp_jan`
*   `temp_jul`

```python
import statsmodels.formula.api as smf

dependent_var = 'Total_sp_log'
# Usar todas as variáveis preditoras na regressão múltipla
multiple_predictor_vars = [
    'Altitude',
    'Actual_river_slope_log',
    'Elevation_mean_catch',
    'prec_ann_catch_log',
    'temp_ann',
    'temp_jan',
    'temp_jul'
]

# Construir a fórmula para a regressão múltipla
formula_multiple = f'{dependent_var} ~ ' + ' + '.join(multiple_predictor_vars)

print(f"Fórmula do modelo de regressão múltipla: {formula_multiple}")

# Ajustar o modelo de regressão linear múltipla
multiple_model = smf.ols(formula=formula_multiple, data=df).fit()

print("\n--- Sumário do Modelo de Regressão Linear Múltipla ---")
print(multiple_model.summary())

# Comparar os coeficientes com os modelos simples
print("\n--- Comparação dos Coeficientes (Modelo Simples vs. Modelo Múltiplo) ---")
print(pd.DataFrame({
    'Coeficiente_Simples': {pred: simple_models[pred].params[pred] for pred in multiple_predictor_vars},
    'Coeficiente_Multiplo': multiple_model.params[multiple_predictor_vars]
}))
```

### Gráficos de Dependência Parcial (Partial Dependence Plots)

Para visualizar o efeito de cada preditor no modelo de regressão múltipla, controlando o efeito das outras variáveis, podemos gerar gráficos de dependência parcial. Como `statsmodels` não possui uma funcionalidade nativa de Partial Dependence Plots (PDPs) tão direta quanto algumas outras bibliotecas (como `scikit-learn` ou `ELI5`), podemos simular um efeito similar fixando as outras variáveis em seus valores médios (ou medianos) para cada preditor. No entanto, para visualizações mais claras em regressão linear, o `sm.graphics.plot_fit` pode ser usado para ver o efeito de cada preditor individualmente no modelo, o que já foi feito nas regressões simples. Para o modelo múltiplo, as visualizações de efeitos parciais são mais complexas. Uma alternativa mais simples é plotar o valor predito em função de uma variável, mantendo as outras constantes. No entanto, o `statsmodels` oferece o `sm.graphics.plot_fit` para mostrar como o modelo se ajusta a uma variável específica, mantendo o restante do modelo fixo. Vamos usar isso para cada variável preditora.

```python
import statsmodels.api as sm

print("\n--- Gráficos de Dependência Parcial (Visualização dos Efeitos no Modelo Múltiplo) ---")

for pred_var in multiple_predictor_vars:
    fig, ax = plt.subplots(figsize=(8, 6))
    fig = sm.graphics.plot_fit(multiple_model, pred_var, ax=ax)
    ax.set_ylabel(dependent_var)
    ax.set_xlabel(pred_var)
    ax.set_title(f'Efeito Parcial de {pred_var} no Modelo Múltiplo')
    plt.show()
```
